# Strings in Python

Strings are sentences or words that are enclosed in either single quotes (' '), double quotes (" "), or triple quotes (''' ''' or """ """). They are used to represent text in Python.

```python
greeting = "Hello, World!"
print(greeting)
print(type(greeting))
```

### Assigning Strings
Assign a string to a variable.

```python
message = "Hello Abhinav"
print(message)
```

### Length of a String
Use `len()` to get the total number of characters in a string.

```python
print(len(message))
```

### String Indexing
Access individual characters using indexes. Positive indexing starts from `0`, while negative indexing starts from `-1` (last character).

```python
print(message[1])   # Positive Indexing
print(message[-2])  # Negative Indexing
```

### String Slicing
Extract a range of characters using slicing `[start:end]`. The `start` index is inclusive, and the `end` index is exclusive.

```python
print(message[0:5])  # From index 0 to 4
print(message[:5])   # Implicitly starts at 0
print(message[6:])   # From index 6 to the end
```

### String Methods: lower() and upper()
Convert the entire string to lowercase using `lower()` or to uppercase using `upper()`.

```python
print(message.lower())
print(message.upper())
```

### String Methods: count()
Use `count()` to find the number of occurrences of a specific substring or character within the string.

```python
print(message.count("Hello"))
print(message.count("l"))
```

### String Methods: find()
Use `find()` to get the starting index of the first occurrence of a substring. It returns `-1` if the substring is not found.

```python
print(message.find('Abhinav'))
```

### String Methods: replace()
Use `replace(old, new)` to replace occurrences of a substring with a new string. Note: Strings are immutable, so this returns a new string and does not modify the original in place.

```python
message = 'Hello Abhinav'
new_message = message.replace('Abhinav', 'Shreya')
print(new_message)
```

### String Concatenation
Combine strings together using the `+` operator.

```python
greeting = 'Hello'
name = 'Corey'
message = greeting + ', ' + name + '. Welcome!'
print(message)
```

### String Formatting: format()
Use `{}` placeholders within a string and call `.format()` to substitute values into the placeholders.

```python
message = '{}, {}. Welcome !!'.format(greeting, name)
print(message)
```

### String Formatting: f-strings (Python 3.6+)
Prefix the string with `f` to directly embed variables or expressions inside `{}`. This is highly readable and allows calling methods directly.

```python
message_f = f'{greeting}, {name}. Welcome!!'
print(message_f)

message_f = f'{greeting}, {name.upper()}. Welcome!!'
print(message_f)
```

### The dir() Function
Pass a variable to `dir()` to see a list of all available attributes and methods for that object type.

```python
print(dir(name))
```

### The help() Function
Pass an object type (like `str`) or method to `help()` to view detailed documentation on its usage and functionality.

```python
print(help(str))
```
